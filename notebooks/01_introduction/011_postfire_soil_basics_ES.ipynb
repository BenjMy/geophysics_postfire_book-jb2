# Restauración del suelo y el ciclo del agua tras incendios forestales

Los incendios forestales tienen efectos profundos y duraderos sobre las propiedades físicas, químicas y biológicas del suelo. Comprender estos efectos es fundamental para diseñar estrategias de restauración eficaces y monitorizar la recuperación del ecosistema a lo largo del tiempo {cite}`lucas-borja_changes_2021`.

## Estrategias de restauración

Tras un incendio forestal se aplican tres enfoques de intervención principales, en función de la pendiente, la severidad y los objetivos de restauración {cite}`lucas-borja_changes_2021`:

| Estrategia | Mecanismo | Aplicación típica |
|---|---|---|
| **Mulching** | Protege la superficie del suelo del impacto de las gotas de lluvia; retiene la humedad | Pendientes suaves a moderadas, fase inicial post-incendio |
| **Barreras de retención / troncos** | Ralentiza la escorrentía superficial; retiene sedimentos | Pendientes pronunciadas, barrancos |
| **Suelo desnudo** (control) | Sin intervención; referencia de recuperación natural | Parcelas de seguimiento y comparación |

```{tip} Elección de la estrategia adecuada
La elección de la estrategia depende del tiempo transcurrido desde el incendio (véase [Cronograma de actuación](#cronograma-de-actuacion)), la pendiente, el régimen de precipitaciones y el ecosistema objetivo. La combinación de mulching con barreras de erosión ha mostrado efectos sinérgicos en ambientes mediterráneos {cite}`lucas-borja_changes_2021`.
```

### Ilustraciones de campo

Las siguientes imágenes muestran condiciones típicas de campo tras un incendio y las técnicas de intervención.

```{figure} ../../assets/images/errosion_barriers.jpg
:name: fig-erosion-barriers
:width: 70%
:align: center

Barreras de troncos instaladas en una ladera quemada para retener sedimentos y ralentizar la escorrentía superficial (Crédito: B. Mary, proyecto GRwater).
```

## Cronograma de actuación

Las herramientas de restauración deben desplegarse en función de la fase de regeneración post-incendio. Las tres ventanas críticas son:

```{list-table} Intervenciones recomendadas por fase post-incendio
:header-rows: 1
:name: table-timeline

* - Fase
  - Momento
  - Acciones prioritarias
* - **Emergencia**
  - Inmediatamente tras el incendio
  - Barreras de erosión, mulching en pendientes pronunciadas, aplicación de semillas
* - **Recuperación temprana**
  - ~3 años post-incendio
  - Mantenimiento de barreras, plantación selectiva, seguimiento de la repelencia al agua
* - **Recuperación tardía**
  - ~8 años post-incendio
  - Retirada de barreras si la vegetación está establecida; evaluación del secuestro de carbono a largo plazo
```

```{warning} Perspectiva a largo plazo
La recuperación completa del ecosistema tras un incendio forestal severo puede llevar décadas. La ventana de 8 años recoge la transición de la restauración activa a la sucesión natural {cite}`lucas-borja_fostering_2024`.
```

## El marco de la Zona Crítica

La restauración del suelo tras incendios se comprende mejor dentro del marco de la **Zona Crítica** (ZC), que considera la delgada capa de la Tierra desde la **parte superior del dosel vegetal hasta la roca madre no alterada** como un sistema único e integrado {cite}`kopp_perspectives_2023`.

```{figure} ../../assets/images/ECZ.jpg
:name: fig-critical-zone
:width: 50%
:align: center
Figura de {cite}`kopp_perspectives_2023`: la zona crítica a escalas anidadas, desde el paisaje hasta el nivel microscópico. El proyecto GRwater investiga los flujos de agua desde la escala de paisaje hasta la escala local en relación con la biodiversidad y la restauración de ecosistemas.
```

```{admonition} ¿Qué es la Zona Crítica?
:class: tip
La Zona Crítica es la delgada envoltura de la Tierra donde la roca, el suelo, el agua, el aire y los organismos vivos interactúan para sostener la vida — la mismísima piel de nuestro planeta, que produce alimentos, filtra el agua y almacena carbono.
```

El concepto de ZC surgió como un **cambio de paradigma** que superó los compartimentos disciplinarios, donde los científicos de distintos campos trabajaban en gran medida de forma aislada:

- Los **pedólogos** estudiaban el suelo como un sustrato estático de crecimiento
- Los **hidrólogos** trataban la infiltración como una propiedad puramente hidráulica
- Los **ecólogos** examinaban la biota de forma independiente al sustrato físico
- Los **geólogos** se centraban en la meteorización de la roca desconectada de los procesos superficiales

Hoy en día, estos se entienden como **procesos profundamente acoplados** que operan a lo largo de un continuo de escalas espaciales y temporales — la formación del suelo, el ciclo del agua, el almacenamiento de carbono y la actividad biológica no pueden separarse de manera significativa.

```{admonition} ¿Por qué importa esto para la restauración post-incendio?
:class: important
Un enfoque puramente agronómico podría centrarse en la capa superficial del suelo con resiembra o enmiendas orgánicas. La perspectiva de la ZC plantea una pregunta más exigente: **¿cómo ha reestructurado el fuego todo el perfil de meteorización?** Desde la destrucción de la interceptación del dosel en superficie hasta los cambios en la recarga de aguas subterráneas en profundidad, la recuperación debe entenderse como una respuesta del sistema, no como una corrección superficial.
```

Los métodos geofísicos son especialmente adecuados para este marco porque pueden **caracterizar la estructura de la ZC de forma no invasiva** a las escalas en que operan estos procesos acoplados — cerrando la brecha entre las mediciones puntuales (testigos de suelo, ensayos de infiltración) y los patrones a escala de paisaje que gobiernan las trayectorias de recuperación post-incendio {cite}`parsekian_multiscale_2015`.

---

```{tip} Resumen
En esta introducción has aprendido:
- Las **principales amenazas** para el suelo tras un incendio forestal (erosión, pérdida de materia orgánica, escorrentía, alteración del flujo de agua).
- Las tres principales **estrategias de restauración** y cuándo aplicarlas.
- El **cronograma en tres fases** (1, 3, 8 años) que guía las decisiones de seguimiento e intervención.
- Cómo el **marco de la Zona Crítica** proporciona una visión holística de la dinámica del suelo tras el incendio.
```
