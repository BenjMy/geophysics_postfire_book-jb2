# Post-fire soil and water cycle restoration

Wildfires have profound and lasting effects on soil physical, chemical, and biological properties. Understanding these effects is critical for designing effective restoration strategies and monitoring ecosystem recovery over time {cite}`lucas-borja_changes_2021`.

## Remediation Strategies

Three broad intervention approaches are commonly applied after wildfires, depending on slope, severity, and restoration goals {cite}`lucas-borja_changes_2021`:

| Strategy | Mechanism | Typical application |
|---|---|---|
| **Mulching** | Protects soil surface from raindrop impact; retains moisture | Gentle to moderate slopes, early post-fire |
| **Erosion logs / barriers** | Slows overland flow; traps sediment | Steep slopes, gullies |
| **Bare soil** (control) | No intervention; natural recovery reference | Monitoring and comparison plots |

```{tip} Choosing the right strategy
The choice of strategy depends on the time elapsed since the fire (see [Timeline of Actuation](#timeline-of-actuation)), slope gradient, rainfall regime, and the target ecosystem. Combining mulching with erosion barriers has shown synergistic effects in Mediterranean environments {cite}`lucas-borja_changes_2021`.
```




### Field Illustrations

The following images show typical post-fire field conditions and intervention techniques.

```{figure} ../../assets/images/errosion_barriers.jpg
:name: fig-erosion-barriers
:width: 70%
:align: center

Erosion log barriers installed on a burned slope to trap sediment and slow overland flow (Credit: B. Mary, GRwater project).
```

## Timeline of actuation

Restoration tools should be deployed according to the phase of post-fire regeneration. The three critical windows are:

```{list-table} Recommended interventions by post-fire phase
:header-rows: 1
:name: table-timeline

* - Phase
  - Timing
  - Priority actions
* - **Emergency**
  - Immediately after fire
  - Erosion barriers, mulching on steep slopes, seed application
* - **Early recovery**
  - ~3 years post-fire
  - Maintain barriers, selective planting, monitor water repellency
* - **Late recovery**
  - ~8 years post-fire
  - Remove barriers if vegetation established; assess long-term C sequestration
```

```{warning} Long-term perspective
Full ecosystem recovery after severe wildfire can take decades. The 8-year window captures the transition from active restoration to natural succession {cite}`lucas-borja_fostering_2024`.
```

## The Critical Zone Framework

Post-fire soil restoration is best understood within the **Critical Zone** (CZ) framework, which considers the thin layer of Earth from the **top of the vegetation canopy down to fresh bedrock** as a single, integrated system {cite}`kopp_perspectives_2023`.


```{figure} ../../assets/images/ECZ.jpg
:name: fig-critical-zone
:width: 50%
:align: center
Figure from {cite}`kopp_perspectives_2023`: the critical zone at nested scales from landscape to microscopic. The GRwater project investigates water fluxes from landscape to local scale in relation to biodiversity and ecosystem restoration.
```

```{admonition} What is the Critical Zone?
:class: tip
The Critical Zone is the thin envelope of Earth where rock, soil, water, air, and living organisms interact to sustain life — the very skin of our planet that produces food, filters water, and stores carbon.
```

The CZ concept emerged as a **paradigm shift** away from disciplinary silos, where scientists from different fields worked largely in isolation:

- **Pedologists** studied soil as a static growing medium
- **Hydrologists** treated infiltration as a purely hydraulic property
- **Ecologists** examined biota independently of the physical substrate
- **Geologists** focused on rock weathering disconnected from surface processes

Today, these are understood as **deeply coupled processes** operating across a continuum of spatial and temporal scales — soil formation, water cycling, carbon storage, and biological activity cannot be meaningfully separated.

```{admonition} Why does this matter for post-fire remediation?
:class: important
A purely agronomic approach might target the topsoil with reseeding or organic amendments. A CZ perspective asks a harder question: **how has fire restructured the entire weathering profile?** From the destruction of canopy interception at the surface to changes in groundwater recharge at depth, recovery must be understood as a system response, not a surface fix.
```

Geophysical methods are uniquely suited to this framework because they can **image CZ structure non-invasively** across the scales at which these coupled processes operate — bridging the gap between point measurements (soil cores, infiltration tests) and the landscape-scale patterns that govern post-fire recovery trajectories {cite}`parsekian_multiscale_2015`.


---
```{tip} Summary
In this introduction you have learned:
- The **main threats** to soil after wildfire (erosion, organic matter loss, runoff, altered water flow).
- The three main **remediation strategies** and when to apply them.
- The **three-phase timeline** (1, 3, 8 years) that guides monitoring and intervention decisions.
- How the **Critical Zone framework** provides a holistic view of post-fire soil dynamics.

```
