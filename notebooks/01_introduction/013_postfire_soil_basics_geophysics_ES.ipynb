```{caution} Una nota de cautela
La aplicación de métodos geofísicos a suelos post-incendio es un **campo emergente**. Las sensibilidades indicadas a continuación se basan en principios geofísicos generales y en estudios de casos preliminares — son **hipótesis de trabajo, no hechos establecidos**. Los cambios inducidos por el fuego en las propiedades del suelo son complejos, a menudo concurrentes, y pueden producir señales geofísicas contrapuestas o ambiguas. La validación empírica en suelos quemados sigue siendo limitada.
```

En el marco del proyecto [GRwater (2025-2027)](../practical_cases/introduction_GRWATER.md) en curso, estamos tratando de utilizar métodos geofísicos repitiendo levantamientos geofísicos en distintos años post-incendio en diferentes parcelas de campo. Podemos así seguir la trayectoria temporal de la recuperación del suelo y evaluar la eficacia de diferentes intervenciones de restauración — sin perturbar las parcelas que estamos monitorizando. La [figura conceptual](fig-concept) muestra cómo diseccionar la cuenca afectada por el incendio utilizando diferentes métodos según la densidad arbolada, entre otros factores. Más resultados están disponibles en las secciones [Agramon EM](../practical_cases/Agramon_EM.md), [Agramon ERT](../practical_cases/Agramon_ERT.md) y [Hellin EM](../practical_cases/Hellin_EM.md).

```{figure} ../../assets/images/concept_fig.png
:name: fig-concept
:width: 80%
:align: center
Figura conceptual que ilustra cómo métodos geofísicos complementarios detectan cambios en las vías de humedad subsuperficial, la estructura del suelo y la topografía superficial tras el incendio forestal y los tratamientos de restauración.
```

**Leyenda:**

⚪ Improbable o no demostrada
❓ Teóricamente posible, pero dirección y magnitud desconocidas

| Propiedad del suelo | ERT | EMI | GPR | Sísmica | Cambio hipotético tras el incendio |
|---|---|---|---|---|---|
| Contenido de agua | 🟡 Probablemente alta | 🟡 Probablemente alta | 🟡 Probablemente alta | 🟡 Probablemente alta | Probablemente reducido por hidrofobicidad — pero variable |
| Contenido en arcilla y textura | 🟡 Posiblemente moderada | 🟡 Posiblemente moderada | ⚪ Improbable | ⚪ Improbable | Probablemente sin cambios a corto plazo — aunque la mezcla con cenizas es incierta |
| Densidad aparente / compactación | 🟠 Indirecta, incierta | 🟠 Indirecta, incierta | ⚪ Improbable | 🟡 Posiblemente | Puede aumentar — evidencia limitada en contextos post-incendio |
| Concentración iónica (salinidad) | 🟡 Probablemente sensible | 🟡 Probablemente sensible | ⚪ Improbable | ⚪ Improbable | Puede aumentar con la lixiviación de cenizas — cronología mal delimitada |
| Capa de carbón / ceniza | 🟠 Débilmente, incierta | 🟠 Débilmente, incierta | 🟠 Posiblemente — no probado | ⚪ Improbable | Presente inmediatamente tras el incendio — firma poco clara |
| Pérdida de materia orgánica | 🟠 Posiblemente indirecta | 🟠 Posiblemente indirecta | ⚪ Improbable | ⚪ Improbable | Fuertemente reducida — detectabilidad geofísica desconocida |
| Profundidad hasta la roca madre | 🟡 Moderada | ⚪ Improbable | 🟡 Moderada | 🟡 Probable | Sin cambios — aunque el frente de meteorización puede desplazarse con los años |
| Capa hidrofóbica | 🟠 Especulativa | 🟠 Especulativa | 🟠 Especulativa | ⚪ Improbable | Esperada — no se ha documentado aún ninguna firma geofísica clara |

El fuego transforma fundamentalmente las propiedades físicas del suelo — quemando la materia orgánica, creando capas hidrofóbicas, alterando la estructura de los poros y redistribuyendo la humedad {cite}`alcaniz_effects_2018`. Estos cambios producen **contrastes** medibles que la geofísica puede monitorizar de forma no invasiva en el espacio y en el tiempo {cite}`mary2020coupled`.
